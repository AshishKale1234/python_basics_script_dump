**Integers** → `int` = Whole numbers such as `200`, `2`, `3`, `99`.

**Floating Point** → `float` = Decimal numbers such as `200.3`, `45.6`, `920.09`, `10.0`.

**String** → `str` = Ordered sequence of characters such as `"hello"`, `"Friday"`, `"320"`, `"ありがとう"`.

**List** → `list` = Ordered sequence of objects such as `["hello", 23, "ありがとう"]`.

<mark>In Python, ordered sequence means elements have a fixed positional order, and you can access them using their index.</mark>

**Dictionaries** → `dict` = Insertion-ordered key-value pairs such as `{"mykey1": "myvalue1", "mykey2": "myvalue2"}`.

**Tuples** → `tuple` = Ordered sequence of objects, however immutable, such as `("Hello", 23.0, 45, "ありがとう")`.

**Sets** → `set` = Unordered collection of unique objects such as `{"a", 23.0, 0, 343}`.

**Booleans** → `bool` = Logical value indicating `True` or `False`.

All of the above mentioned are Data types, but not necessarily Data Structures, there is a slight difference in Data types and Data structures.

`Data type`: Defines what kind of value a variable holds.
`Data structure`: Organizes and stores data in a particular way.
