***Rules for Variables***

`Rule 1`: Names of the variables cannot start with a number


`Rule 2`: There can be no spaces in the name, '_' should be used instead


`Rule 3`: Should only contain (A–Z, a–z), digits (0–9), and underscores (_)

***Python is Dynamically typed***, which means the same variable name can be given for a different set of objects, this is unique compared to some other programming languages that are statically typed.
